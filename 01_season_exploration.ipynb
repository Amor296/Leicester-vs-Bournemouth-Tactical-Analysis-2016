## Leicester City 2015/2016: The Miracle Season Exploration
In this notebook, we will connect to the **StatsBomb Open Data** to extract and filter all matches played by Leicester City during their historic Premier League winning campaign.

```python
# Import necessary libraries
from statsbombpy import sb
import pandas as pd

# Fetch all available competitions from StatsBomb
all_competitions = sb.competitions()

# Filter for Premier League 2015/2016 season
# Competition ID for PL is 2, Season ID for 15/16 is 27
target_season = all_competitions[(all_competitions.competition_name == 'Premier League') & 
                                 (all_competitions.season_name == '2015/2016')]
target_season
```

## Extracting Leicester City Matches
Now we will retrieve the full list of matches for this season and isolate the fixtures where **Leicester City** was either the home or away team.

```python
# Load all match fixtures for the 2015/2016 PL season
all_matches = sb.matches(competition_id=2, season_id=27)

# Filter fixtures to only include Leicester City matches
leicester_fixtures = all_matches[(all_matches.home_team == 'Leicester City') | 
                                 (all_matches.away_team == 'Leicester City')]

# Sort the matches by date to follow the season timeline
leicester_fixtures = leicester_fixtures.sort_values('match_date')

# Display the first 10 matches of the season
leicester_fixtures[['match_date', 'home_team', 'away_team', 'home_score', 'away_score']].head(10)
```

## Identifying Key "Big Six" Victories
To analyze the tactics of an underdog, we must look at how they performed against the league giants. We will filter for matches against the "Big Six" to find high-value tactical data.

```python
# Define a list of traditional 'Big Six' opponents
big_six = ['Manchester City', 'Chelsea', 'Liverpool', 'Arsenal', 'Manchester United', 'Tottenham Hotspur']

# Filter for matches where Leicester played against these teams
tactical_highlights = leicester_fixtures[
    (leicester_fixtures.home_team.isin(big_six)) | 
    (leicester_fixtures.away_team.isin(big_six))
]

# Print summary and show the specific high-interest matches
print(f"Total Leicester matches in 2015/16: {len(leicester_fixtures)}")
tactical_highlights[['match_id', 'match_date', 'home_team', 'away_team', 'home_score', 'away_score']]
```
